### Importação dos dados



```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja1 = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

print("Loja 1:")
print(loja1.head())
print("\nLoja 2:")
print(loja2.head())
print("\nLoja 3:")
print(loja3.head())
print("\nLoja 4:")
print(loja4.head())
```

#1. Análise do faturamento


```python
faturamento_loja1 = loja1['Preço'].sum()
faturamento_loja2 = loja2['Preço'].sum()
faturamento_loja3 = loja3['Preço'].sum()
faturamento_loja4 = loja4['Preço'].sum()

faturamento_por_loja = {
    'Loja 1': faturamento_loja1,
    'Loja 2': faturamento_loja2,
    'Loja 3': faturamento_loja3,
    'Loja 4': faturamento_loja4
}

faturamento_df = pd.DataFrame(list(faturamento_por_loja.items()), columns=['Loja', 'Faturamento Total'])
print("\nFaturamento Total por Loja (Corrigido):")
print(faturamento_df)

plt.figure(figsize=(8, 6))
sns.barplot(x='Loja', y='Faturamento Total', data=faturamento_df)
plt.title('Faturamento Total por Loja (Corrigido)')
plt.xlabel('Loja')
plt.ylabel('Faturamento Total')
plt.show()
```

# 2. Vendas por Categoria


```python
def top_categorias(df, nome_loja):
    contagem_categorias = df['Categoria do Produto'].value_counts().nlargest(5)
    print(f"\nTop 5 Categorias da {nome_loja}:")
    print(contagem_categorias)
    contagem_categorias.plot(kind='bar')
    plt.title(f'Top 5 Categorias - {nome_loja}')
    plt.xlabel('Categoria do Produto')
    plt.ylabel('Número de Vendas')
    plt.show()

top_categorias(loja1, 'Loja 1')
top_categorias(loja2, 'Loja 2')
top_categorias(loja3, 'Loja 3')
top_categorias(loja4, 'Loja 4')
```

# 3. Média de Avaliação das Lojas

```python
def distribuicao_avaliacoes(df, nome_loja):
    plt.figure(figsize=(8, 6))
    sns.histplot(df['Avaliação da compra'], bins=5, discrete=True)
    plt.title(f'Distribuição das Avaliações - {nome_loja}')
    plt.xlabel('Avaliação (1-5)')
    plt.ylabel('Número de Avaliações')
    plt.xticks(range(1, 6))
    plt.show()
    print(f"\nEstatísticas das Avaliações - {nome_loja}:")
    print(df['Avaliação da compra'].describe())

distribuicao_avaliacoes(loja1, 'Loja 1')
distribuicao_avaliacoes(loja2, 'Loja 2')
distribuicao_avaliacoes(loja3, 'Loja 3')
distribuicao_avaliacoes(loja4, 'Loja 4')
```

# 4. Produtos Mais e Menos Vendidos

```python

def top_produtos(df, nome_loja, n=5):
    contagem_produtos = df['Produto'].value_counts().nlargest(n)
    print(f"\nTop {n} Produtos Mais Vendidos da {nome_loja}:")
    print(contagem_produtos)
    contagem_produtos.plot(kind='bar')
    plt.title(f'Top {n} Produtos Mais Vendidos - {nome_loja}')
    plt.xlabel('Produto')
    plt.ylabel('Número de Vendas')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

top_produtos(loja1, 'Loja 1')
top_produtos(loja2, 'Loja 2')
top_produtos(loja3, 'Loja 3')
top_produtos(loja4, 'Loja 4')
```

```python
def bottom_produtos(df, nome_loja, n=5):
    contagem_produtos = df['Produto'].value_counts().nsmallest(n)
    print(f"\nTop {n} Produtos Menos Vendidos da {nome_loja}:")
    print(contagem_produtos)
    contagem_produtos.plot(kind='bar')
    plt.title(f'Top {n} Produtos Menos Vendidos - {nome_loja}')
    plt.xlabel('Produto')
    plt.ylabel('Número de Vendas')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

bottom_produtos(loja1, 'Loja 1')
bottom_produtos(loja2, 'Loja 2')
bottom_produtos(loja3, 'Loja 3')
bottom_produtos(loja4, 'Loja 4')
```

# 5. Frete Médio por Loja


```python
frete_medio_loja1 = loja1['Frete'].mean()
frete_medio_loja2 = loja2['Frete'].mean()
frete_medio_loja3 = loja3['Frete'].mean()
frete_medio_loja4 = loja4['Frete'].mean()

frete_medio_por_loja = {
    'Loja 1': frete_medio_loja1,
    'Loja 2': frete_medio_loja2,
    'Loja 3': frete_medio_loja3,
    'Loja 4': frete_medio_loja4
}

frete_medio_df = pd.DataFrame(list(frete_medio_por_loja.items()), columns=['Loja', 'Frete Médio'])
print("\nFrete Médio por Loja:")
print(frete_medio_df)

plt.figure(figsize=(8, 6))
sns.barplot(x='Loja', y='Frete Médio', data=frete_medio_df)
plt.title('Frete Médio por Loja')
plt.xlabel('Loja')
plt.ylabel('Frete Médio')
plt.show()
```

# Gráficos

```python
import numpy as np

# Dados de faturamento (já calculados anteriormente)
faturamento_por_loja_dict = {
    'Loja 1': faturamento_loja1,
    'Loja 2': faturamento_loja2,
    'Loja 3': faturamento_loja3,
    'Loja 4': faturamento_loja4
}
faturamento_df = pd.DataFrame(list(faturamento_por_loja_dict.items()), columns=['Loja', 'Faturamento Total'])

# Dados de média de avaliação (já calculados anteriormente)
media_avaliacao_por_loja_dict = {
    'Loja 1': media_avaliacao_loja1,
    'Loja 2': media_avaliacao_loja2,
    'Loja 3': media_avaliacao_loja3,
    'Loja 4': media_avaliacao_loja4
}
media_avaliacao_df = pd.DataFrame(list(media_avaliacao_por_loja_dict.items()), columns=['Loja', 'Média da Avaliação'])

# Merge dos DataFrames para facilitar a plotagem
comparativo_df = pd.merge(faturamento_df, media_avaliacao_df, on='Loja')

# Configuração do gráfico
labels = comparativo_df['Loja']
faturamento = comparativo_df['Faturamento Total']
avaliacoes = comparativo_df['Média da Avaliação']

x = np.arange(len(labels))  # Localização das barras
width = 0.35  # Largura das barras

fig, ax1 = plt.subplots(figsize=(10, 6))

rects1 = ax1.bar(x - width/2, faturamento, width, label='Faturamento Total', color='skyblue')
ax2 = ax1.twinx()  # Segundo eixo y para a média de avaliação
rects2 = ax2.bar(x + width/2, avaliacoes * max(faturamento) / 5, width, label='Média da Avaliação (Escal.)', color='salmon') # Escala para visualização conjunta

# Adicionando rótulos e títulos
ax1.set_ylabel('Faturamento Total', color='skyblue')
ax2.set_ylabel('Média da Avaliação (Escal.)', color='salmon')
ax1.set_xlabel('Loja')
ax1.set_title('Comparativo de Faturamento Total e Média de Avaliação por Loja')
ax1.set_xticks(x)
ax1.set_xticklabels(labels)
ax1.legend(loc='upper left')
ax2.legend(loc='upper right')

fig.tight_layout()
plt.show()
```

```python
# Calcular o número de vendas por produto e a avaliação média por produto na Loja 1
vendas_por_produto_loja1 = loja1['Produto'].value_counts().reset_index()
vendas_por_produto_loja1.columns = ['Produto', 'Número de Vendas']

avaliacao_por_produto_loja1 = loja1.groupby('Produto')['Avaliação da compra'].mean().reset_index()
avaliacao_por_produto_loja1.columns = ['Produto', 'Avaliação Média']

# Merge dos DataFrames
relacao_produto_avaliacao_loja1 = pd.merge(vendas_por_produto_loja1, avaliacao_por_produto_loja1, on='Produto')

# Gráfico de dispersão
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Número de Vendas', y='Avaliação Média', data=relacao_produto_avaliacao_loja1)
plt.title('Relação entre Número de Vendas e Avaliação Média (Loja 1)')
plt.xlabel('Número de Vendas')
plt.ylabel('Avaliação Média')
plt.grid(True)
plt.show()
```

```python
# Calcular o faturamento acumulado para a Loja 4
faturamento_acumulado_loja4 = loja4['Preço'].cumsum()

# Criar um índice para o eixo x (assumindo ordem temporal)
indices = range(len(faturamento_acumulado_loja4))

# Gráfico de linha
plt.figure(figsize=(10, 6))
plt.plot(indices, faturamento_acumulado_loja4, marker='o', linestyle='-', color='purple')
plt.title('Tendência de Faturamento Acumulado ao Longo das Transações (Loja 4)')
plt.xlabel('Número da Transação (Ordem Implícita)')
plt.ylabel('Faturamento Acumulado')
plt.grid(True)
plt.tight_layout()
plt.show()
```
